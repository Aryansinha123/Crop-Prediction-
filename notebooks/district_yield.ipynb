```python
import pandas as pd

# Load original dataset
df = pd.read_csv("../datasets/yield/crop_production.csv")

# Filter Uttar Pradesh + Wheat for historical years (2005 - 2014)
df = df[
    (df['State_Name'] == 'Uttar Pradesh') &
    (df['Crop'] == 'Wheat') &
    (df['Crop_Year'] >= 2005) &
    (df['Crop_Year'] <= 2014)
].copy()

# Create Yield column
df['Yield'] = df['Production'] / df['Area']

# Standardize district names to match other files (Title Case)
df['District_Name'] = df['District_Name'].str.title()

# Map district names to align with weather and coordinates datasets
df.loc[df['District_Name'] == 'Allahabad', 'District_Name'] = 'Prayagraj'
df.loc[df['District_Name'] == 'Kanpur Nagar', 'District_Name'] = 'Kanpur'

# Keep only the 10 districts in target list
districts = ['Lucknow', 'Kanpur', 'Agra', 'Varanasi', 'Prayagraj', 'Bareilly', 'Gorakhpur', 'Meerut', 'Aligarh', 'Jhansi']
df = df[df['District_Name'].isin(districts)].copy()

# Keep required columns
district_yield = df[[
    'District_Name',
    'Crop_Year',
    'Yield'
]].copy()

# Rename Crop_Year to Year
district_yield.rename(
    columns={'Crop_Year': 'Year'},
    inplace=True
)

# Shift years by 10 to align with 2015-2024 satellite and weather data
district_yield['Year'] = district_yield['Year'] + 10

# Save CSV
district_yield.to_csv(
    "../datasets/yield/district_yield.csv",
    index=False
)

print("district_yield.csv created successfully!")
print("Total rows:", len(district_yield))
print(district_yield.head(10))
```
