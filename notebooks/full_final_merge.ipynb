```python
import pandas as pd

# ==========================================
# LOAD WEATHER DATA
# ==========================================

weather = pd.read_csv(
    "../datasets/weather/full_up_weather.csv"
)

print("WEATHER DATA")
print(weather.head())

print("\nWeather Rows:")
print(len(weather))


# ==========================================
# LOAD NDVI DATA
# ==========================================

ndvi = pd.read_csv(
    "../datasets/satellite/Full_UP_District_NDVI_2015_2024.csv"
)

print("\nNDVI DATA")
print(ndvi.head())

print("\nNDVI Rows:")
print(len(ndvi))


# ==========================================
# LOAD YIELD DATA
# ==========================================

yield_data = pd.read_csv(
    "../datasets/yield/district_yield.csv"
)

print("\nYIELD DATA")
print(yield_data.head())

print("\nYield Rows:")
print(len(yield_data))


# ==========================================
# CLEAN DISTRICT NAMES
# ==========================================

weather['District_Name'] = (
    weather['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)

ndvi['District_Name'] = (
    ndvi['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)

yield_data['District_Name'] = (
    yield_data['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)


# ==========================================
# ENSURE YEAR IS INTEGER
# ==========================================

weather['Year'] = weather['Year'].astype(int)

ndvi['Year'] = ndvi['Year'].astype(int)

yield_data['Year'] = yield_data['Year'].astype(int)


# ==========================================
# MERGE WEATHER + NDVI
# ==========================================

merged = pd.merge(

    weather,

    ndvi,

    on=['District_Name', 'Year'],

    how='inner'

)

print("\nAFTER WEATHER + NDVI MERGE")

print(merged.head())

print("\nRows:")
print(len(merged))


# ==========================================
# MERGE WITH YIELD
# ==========================================

final_df = pd.merge(

    merged,

    yield_data,

    on=['District_Name', 'Year'],

    how='inner'

)

print("\nFINAL MERGED DATASET")

print(final_df.head())

print("\nFINAL ROW COUNT:")
print(len(final_df))


# ==========================================
# REMOVE INVALID VALUES
# ==========================================

final_df = final_df.dropna()

final_df = final_df[
    final_df['NDVI'] > 0
]


# ==========================================
# SAVE FINAL DATASET
# ==========================================

final_df.to_csv(
    "../datasets/final_large_scale_dataset.csv",
    index=False
)

print("\nFINAL LARGE-SCALE DATASET SAVED!")
```
