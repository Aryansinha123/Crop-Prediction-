```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder


# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    "../datasets/final_large_scale_dataset.csv"
)

print("ORIGINAL DATASET:")
print(df.head())

print("\nTOTAL ROWS:")
print(len(df))


# ==========================================
# SORT DATA
# ==========================================

df = df.sort_values(
    by=['District_Name', 'Year']
)


# ==========================================
# TEMPERATURE ANOMALY
# ==========================================

df['Temp_Anomaly'] = (

    df['T2M']

    -

    df['T2M'].mean()

)

# ==========================================
# RAINFALL ANOMALY
# ==========================================

df['Rain_Anomaly'] = (

    df['PRECTOTCORR']

    -

    df['PRECTOTCORR'].mean()

)

# ==========================================
# HUMIDITY ANOMALY
# ==========================================

df['Humidity_Anomaly'] = (

    df['RH2M']

    -

    df['RH2M'].mean()

)

# ==========================================
# CLIMATE ANOMALY SCORE
# ==========================================

df['Climate_Anomaly_Score'] = (

    abs(df['Temp_Anomaly'])

    +

    abs(df['Rain_Anomaly'] / 100)

    +

    abs(df['Humidity_Anomaly'])

)

# ==========================================
# PREVIOUS YEAR YIELD
# ==========================================

df['Previous_Year_Yield'] = (

    df.groupby('District_Name')['Yield']

    .shift(1)

)

# ==========================================
# ROLLING TEMPERATURE
# ==========================================

df['Rolling_Temp_Avg'] = (

    df.groupby('District_Name')['T2M']

    .transform(

        lambda x: x.rolling(
            window=3,
            min_periods=1
        ).mean()

    )

)

# ==========================================
# ROLLING RAINFALL
# ==========================================

df['Rolling_Rainfall_Avg'] = (

    df.groupby('District_Name')['PRECTOTCORR']

    .transform(

        lambda x: x.rolling(
            window=3,
            min_periods=1
        ).mean()

    )

)

# ==========================================
# DISTRICT ENCODING
# ==========================================

encoder = LabelEncoder()

df['District_Encoded'] = encoder.fit_transform(
    df['District_Name']
)

# ==========================================
# REMOVE NULLS
# ==========================================

df = df.dropna()

# ==========================================
# SAVE FEATURE ENGINEERED DATASET
# ==========================================

df.to_csv(

    "../datasets/feature_engineered_dataset.csv",

    index=False

)

print("\nFEATURE ENGINEERING COMPLETED!")

print("\nNEW DATASET:")
print(df.head())

print("\nFINAL SHAPE:")
print(df.shape)
```
