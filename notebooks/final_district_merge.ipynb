```python
import pandas as pd

# ==========================================
# LOAD DISTRICT WEATHER
# ==========================================

weather = pd.read_csv(
    "../datasets/weather/district_weather.csv"
)

print("\nWEATHER DATA")
print(weather.head())


# ==========================================
# LOAD DISTRICT NDVI
# ==========================================

ndvi = pd.read_csv(
    "../datasets/satellite/district_ndvi.csv"
)

print("\nNDVI DATA")
print(ndvi.head())


# ==========================================
# LOAD DISTRICT YIELD
# ==========================================

yield_data = pd.read_csv(
    "../datasets/yield/district_yield.csv"
)

print("\nYIELD DATA")
print(yield_data.head())


# ==========================================
# CLEAN NDVI DATA
# ==========================================

# Keep only required columns
ndvi = ndvi[['District_Name', 'Year', 'NDVI']]

# Convert Year to int
ndvi['Year'] = ndvi['Year'].astype(float).astype(int)

# Clean district names
ndvi['District_Name'] = ndvi['District_Name'].str.strip()

print("\nCLEANED NDVI")
print(ndvi.head())


# ==========================================
# CLEAN WEATHER DATA
# ==========================================

weather['Year'] = weather['Year'].astype(int)

weather['District_Name'] = weather['District_Name'].str.strip()

print("\nCLEANED WEATHER")
print(weather.head())


# ==========================================
# CLEAN YIELD DATA
# ==========================================

yield_data['Year'] = yield_data['Year'].astype(int)

yield_data['District_Name'] = yield_data['District_Name'].str.strip()

print("\nCLEANED YIELD")
print(yield_data.head())


# ==========================================
# MERGE WEATHER + NDVI
# ==========================================

merged = pd.merge(
    weather,
    ndvi,
    on=['District_Name', 'Year'],
    how='inner'
)

print("\nMERGED WEATHER + NDVI")
print(merged.head())


# ==========================================
# FINAL MERGE WITH YIELD
# ==========================================

final_df = pd.merge(
    merged,
    yield_data,
    on=['District_Name', 'Year'],
    how='inner'
)

# Impute missing NDVI values using district-level mean
final_df['NDVI'] = final_df.groupby('District_Name')['NDVI'].transform(
    lambda x: x.fillna(x.mean())
)

print("\nFINAL DATASET")
print(final_df.head())

print("\nTOTAL ROWS:")
print(len(final_df))


# ==========================================
# SAVE FINAL DATASET
# ==========================================

final_df.to_csv(
    "../datasets/final_district_dataset.csv",
    index=False
)

print("\nFINAL DISTRICT DATASET SAVED!")
```
