```python
import pandas as pd

# ==========================================
# LOAD WEATHER DATA
# ==========================================

weather_df = pd.read_csv(
    "../datasets/weather/full_up_weather.csv"
)

# ==========================================
# LOAD NDVI DATA
# ==========================================

ndvi_df = pd.read_csv(
    "../datasets/satellite/Full_UP_District_NDVI_2015_2024.csv"
)

# ==========================================
# LOAD YIELD DATA
# ==========================================

yield_df = pd.read_csv(
    "../datasets/yield/full_up_yield.csv"
)

# ==========================================
# CLEAN COLUMN NAMES
# ==========================================

weather_df.columns = weather_df.columns.str.strip()
ndvi_df.columns = ndvi_df.columns.str.strip()
yield_df.columns = yield_df.columns.str.strip()

# ==========================================
# CLEAN DISTRICT NAMES
# ==========================================

weather_df['District_Name'] = (
    weather_df['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)

ndvi_df['District_Name'] = (
    ndvi_df['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)

yield_df['District_Name'] = (
    yield_df['District_Name']
    .astype(str)
    .str.strip()
    .str.lower()
)

# ==========================================
# FIX YEAR TYPES
# ==========================================

weather_df['Year'] = weather_df['Year'].astype(int)

ndvi_df['Year'] = (
    ndvi_df['Year']
    .astype(float)
    .astype(int)
)

yield_df['Year'] = yield_df['Year'].astype(int)

# ==========================================
# REMOVE UNUSED NDVI COLUMNS
# ==========================================

drop_cols = []

for col in ndvi_df.columns:

    if (
        'system' in col.lower()
        or '.geo' in col.lower()
    ):
        drop_cols.append(col)

ndvi_df = ndvi_df.drop(columns=drop_cols)

# ==========================================
# PRINT DATA INFO
# ==========================================

print("WEATHER ROWS:", len(weather_df))
print("NDVI ROWS:", len(ndvi_df))
print("YIELD ROWS:", len(yield_df))

# ==========================================
# MERGE WEATHER + NDVI
# ==========================================

merged_df = pd.merge(

    weather_df,

    ndvi_df,

    on=['District_Name', 'Year'],

    how='inner'

)

print("\nAFTER WEATHER + NDVI MERGE:")
print(len(merged_df))

# ==========================================
# MERGE WITH YIELD
# ==========================================

final_df = pd.merge(

    merged_df,

    yield_df,

    on=['District_Name', 'Year'],

    how='inner'

)

print("\nFINAL MERGED ROWS:")
print(len(final_df))

# ==========================================
# REMOVE NULLS
# ==========================================

final_df = final_df.dropna()

# ==========================================
# KEEP VALID NDVI
# ==========================================

final_df = final_df[
    final_df['NDVI'] > 0
]

# ==========================================
# SAVE FINAL DATASET
# ==========================================

final_df.to_csv(

    "../datasets/final_large_scale_dataset.csv",

    index=False

)

print("\nFINAL LARGE SCALE DATASET SAVED!")

print(final_df.head())
```
